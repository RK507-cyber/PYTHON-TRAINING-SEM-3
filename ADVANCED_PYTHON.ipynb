```python
# Single Level Inheritance

class A:
  def sa(self):
    print('Papa ki Bike')
class B(A): # Inheritance
  def sb(self):
    print('Father & sons bike')
obj=B()
obj.sa()
obj.sb()
```

```python
# Multiple Level Inheritance
class A:
  def sa(self):
    print('Papa ki Bike')
class B(A): # Inheritance
  def sb(self):
    print('Father & sons bike')
class C(B):
  def sc(self):
    print('Happy Family')
obj=C() # Last class contains all inheritances so use it for execution
obj.sa()
obj.sb()
obj.sc()
```

```python
# Multiple Inheritance

class A:
  def sa(self):
    print('Papa ki Bike')
class B:
  def sb(self):
    print('Father & sons bike')
class C(A,B):  # Child can use parent's property
  def sc(self):
    print('Happy Family')
obj=C()
obj.sa()
obj.sb()
obj.sc()
```

```python
# Hierarchal Inheritance

class A:
  def sa(self):
    print('Papa ki Bike')
class B(A):
  def sb(self):
    print('Father & sons bike')
class C(A):
  def sc(self):
    print('Happy Family')
obj=C()
obj.sa()  # C has no attribute B (C has A & C), no connection (B has A & B)
obj.sc()
```

```python
# Hybrid Inheritance

class A:
  def sa(self):
    print('Papa ki Bike')
class B(A):
  def sb(self):
    print('Father & sons bike')
class C(A):
  def sc(self):
    print('Happy Family')
class D(B,C):  # D has all A,B & C (Hierarchal and Multilevel)
  def sd(self):
    print('Yay!!')
obj=D()
obj.sa()
obj.sb()
obj.sc()
obj.sd()
```

```python
# Overloading - Assigning additional tasks , same class,same function
# Overriding -  overriding parent inheritance if child has its own function
# Method overriding - Different classes with parent child connection but same
# function

class Papa:
  def bike(self):
    print('Hero Honda')
class Son(Papa):
  def bike(self):
    print('Yamaha')
obj=Son()
obj.bike()
```
