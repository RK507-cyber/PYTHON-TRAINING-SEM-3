```python
# With statement
with open('myfile.txt','w') as f: # 'with' closes the file after execution
  f.write('Eena\n')
  f.write('Mina\n')
  f.write('Deeka\n')
  print('is file closed ?',f.closed) # f.closed asks if the file is closed or not
print('Now is file closed ?', f.closed)
```

```python
f=open('myfile.txt','r')
print(f.tell()) # Tells position of the cursor
print(f.read(3))
print(f.tell())
print(f.read(14))  # Reads after 3 characters
print(f.tell())
```

```python
# seek shifts the cursor to the required character at that index
d="It's raining cats"
f=open('text.txt','w')
f.write(d)

with open('text.txt','r+') as f:
  fd=f.read()
  print(fd)
  print('Current position',f.tell())
  f.seek(17)
  print('Current position',f.tell())
  f.write(' and dogs')
  f.seek(0)
  fd=f.read()
  print(fd)   # seek does not add at the position it shifts the text and fits the newly added text
```

```python
import os,sys
fn=input('Enter file name:')
if os.path.isfile(fn):
  f=open(fn,'r')
lcount=wcount=ccount=0
for line in f:
  lcount+=1
  ccount+=len(line)   # Character count
  words=line.split()  # splits into words in the form of list
  wcount+=len(words)
print('Lines: ',lcount)
print('Words: ',wcount)
print('Character: ',ccount)
```

```python
import csv
with open('Data.csv','w',newline='') as f:
  w=csv.writer(f)
  w.writerow(['Trends','Max','Souled','Decathalon'])
  n=int(input('Enter the no. of employees: '))
  for i in range(n):
    trends=input('Enter shop no.: ')  # n times 4
    max=input('Enter shop no.: ')
    souled=input('Enter shop no.: ')
    dec=input('Enter shop no.: ')
    w.writerow([trends,max,souled,dec])
print('Data inserted successfully')
```

```python
f=open('/content/Data.csv','r')
r=csv.reader(f)
d=list(r)   # stores data in list
print(d)
for line in d :
  for word in line:
    print(word,'\t',end='')
  print()
```
