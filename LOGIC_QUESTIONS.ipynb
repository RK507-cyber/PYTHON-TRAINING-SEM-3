```python
# For common element
y=[1,2,3]
x=[2,3,4]
z=[3,4,5]
f=[]
for i in y:
  for t in x:
    for d in z:
      if i==t & t==d:
        f.append(d)
print(f)
```

```python
d=[[[1,2],[3,4]],[[5,6],[7,8]]] # Understand working
def fun(m):
  v=m[0][0]
  for row in m:
    for element in row:
      if v < element:
        v=element
  return v
print(fun(d[0]))
```

```python
t={}
def addd(index):
  if index in t:
    t[index]+=1  # t is empty
  else:          # index is key value
    t[index]=1  # Gets addded to t dictionary
addd('Apple')
addd('Mango')
addd('apple')
print(len(t))
```
